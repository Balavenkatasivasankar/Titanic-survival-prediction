```python
categories = df['Sex'].value_counts()
plt.figure(figsize=(10, 6))

sns.countplot(data=df, x='Sex')
plt.title('Sex Distribution')
plt.xlabel('Sex')
plt.ylabel('Freq')

for i in range(len(categories)):
    plt.text(i, categories[i] + 5, categories[i], ha='center')

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.pie(x=df['Sex'].value_counts(), labels=df['Sex'].value_counts().index, autopct="%1.2f%%", explode=[0, 0.1])
plt.title('Sex Distribution')
plt.legend()

plt.show()
```

```python
categories = df['Survived'].value_counts()
plt.figure(figsize=(10, 6))

sns.countplot(data=df, x='Survived')
plt.title('Survived Distribution')
plt.xlabel('Survived')
plt.ylabel('Freq')

for i in range(len(categories)):
    plt.text(i, categories[i] + 3, categories[i], ha='center')

plt.show()
```

```python
df['Pclass'].value_counts()
```

```python
categories = df['Pclass'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Pclass', order=sorted(df['Pclass'].unique()))
plt.title('Pclass Distribution')
plt.xlabel('Pclass')
plt.ylabel('Frequency')

# كتابة القيم فوق الأعمدة
for i, val in enumerate(categories):
    plt.text(i, val + 3, val, ha='center')

plt.show()
```
